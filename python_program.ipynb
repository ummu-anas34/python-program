```python
# A Trader buys 15 items at 2,500 each and sells item at 3,200. Write a program that calculates total cost, total sales, and profit.
Total_item = 15
Cost_price = 2500
Selling_price = 3200

total_cost = Total_item * Cost_price
total_sales = Total_item * Selling_price
profit = total_sales - total_cost

print("Total cost:", total_cost)
print("Total sales:", total_sales)
print("Profit:", profit)
```
